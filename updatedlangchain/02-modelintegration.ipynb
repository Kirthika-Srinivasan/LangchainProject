### Models Integration with OpenAI, Google Gemini and GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
```

### ChatGPT model integration

```python
from langchain.chat_models import init_chat_model
model=init_chat_model("gpt-4.1")

model
```

```python
response = model.invoke("Hello, How are you?")
response.content
```

```python
### ChatOpenAI
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model="gpt-4.1")
response=model.invoke("Hi! Who are you?")
response.content
```

### Google Gemini model integration

```python
model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("Why do parrots talk?")

response.content
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model=ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
response = model.invoke("What's up with you?")

response.content
```

### Groq model integration

```python
model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Hello, Who are you?")
response
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="qwen/qwen3-32b")
response=model.invoke("What's ChatGroq?")
response.content
```

### Ollama model integration

```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="qwen2.5:3b")
messages = [
    (
        "system",
        "You are a helpful assistant that translates English to French. Translate the user sentence.",
    ),
    ("human", "I love programming."),
]
response = model.invoke(messages)

response.content
```

### Streaming

```python
model = ChatOpenAI(model="gpt-4.1")
model.invoke("Write me a 200 words paragraph on AI")
```

```python
for chunk in model.stream("Write me a 200 words paragraph on AI"):
    print(chunk.text, end="|", flush=True)
```

### Batch

```python
responses =  model.batch([
    "Why do parrots have colorful colors?",
    "How do airplanes fly?",
    "What is quantum computing?"
])

for resp in responses:
    print(resp)
```

```python
responses =  model.batch([
    "Why do parrots have colorful colors?",
    "How do airplanes fly?",
    "What is quantum computing?"],
    config={
    'max_concurrency': 5, # Limiting to 5 parallel calls
}
)

for resp in responses:
    print(resp)
```
